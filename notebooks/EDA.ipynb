```python
#outlier checking function
def find_outliers(sr,col):
    Q1=sr[col].quantile(0.25)
    Q3=sr[col].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR
    outlier_iqr1=(sr[col]<lower_bound) | (sr[col]>upper_bound)
    print(f"{col}:{len(sr[outlier_iqr1])}")
    return outlier_iqr1
outliers=df["loan_id"]<-1
for i in num_cols:
    outliers=outliers|(find_outliers(df,i))
```

```python
c=0
for i in outliers:
    if i==True:
        c+=1
print(c)
len(df[outliers])
```

```python
print(len(df[outliers==True]))
#remove outliers from numerical columns
df=df[~(outliers)]
print(len(df[outliers==True]))
```
